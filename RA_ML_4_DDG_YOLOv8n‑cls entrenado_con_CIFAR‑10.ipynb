```python
# ============================================================
# ENTRENAMIENTO COMPLETO DE YOLOv8n-cls EN CIFAR-10
# ============================================================

import os
import numpy as np
from PIL import Image
from tensorflow.keras.datasets import cifar10
from google.colab import drive
drive.mount('/content/drive')


# ------------------------------------------------------------
# 1. Cargar CIFAR-10
# ------------------------------------------------------------
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
y_train = y_train.flatten()
y_test = y_test.flatten()

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

# ------------------------------------------------------------
# 2. Crear estructura de carpetas en TU RUTA
# ------------------------------------------------------------
base_path = "/content/drive/MyDrive/CIFAR-10"
train_path = os.path.join(base_path, "train")
val_path = os.path.join(base_path, "val")

for split_path in [train_path, val_path]:
    os.makedirs(split_path, exist_ok=True)
    for cname in class_names:
        os.makedirs(os.path.join(split_path, cname), exist_ok=True)

# ------------------------------------------------------------
# 3. Guardar imágenes en carpetas
# ------------------------------------------------------------
def save_split(images, labels, split_path, prefix):
    for i, (img, label) in enumerate(zip(images, labels)):
        cname = class_names[label]
        Image.fromarray(img).save(os.path.join(split_path, cname, f"{prefix}_{i}.png"))

save_split(x_train, y_train, train_path, "train")
save_split(x_test, y_test, val_path, "val")

print("Estructura creada correctamente en:", base_path)

# ------------------------------------------------------------
# 4. Instalar Ultralytics
# ------------------------------------------------------------
!pip install ultralytics -q
from ultralytics import YOLO

# ------------------------------------------------------------
# 5. Entrenar YOLOv8n-cls (SIN YAML)
# ------------------------------------------------------------
model = YOLO("yolov8n-cls.pt")

results = model.train(
    data=base_path,
    epochs=5, #antes 10
    imgsz=16, #antes 64
    batch=16 #antes 64
)

# ------------------------------------------------------------
# 6. Evaluación final
# ------------------------------------------------------------
metrics = model.val()
print(metrics)
```
